#### Actividad 3 — Estructuras de datos

````python

ventas = [
    {"producto": "Lámpara de mesa", "precio": 105.10, "cantidad": 5},
    {"producto": "Secadora", "precio": 97.96, "cantidad": 3},
    {"producto": "Cuadro decorativo", "precio": 69.94, "cantidad": 5},
    {"producto": "Heladera", "precio": 114.35, "cantidad": 2},
]

total = 0
for venta in ventas:
    subtotal = venta["precio"] * venta["cantidad"]
    print(venta["producto"], "-> $", round(subtotal, 2))
    total = total + subtotal

print("Total de ventas: $", round(total, 2))
````

**¿Listas o diccionarios?**

Se decidió combinar ambas estructuras:

* **Diccionarios para cada venta**, porque permiten acceder a cada dato por su nombre (`venta["precio"]`) en lugar de por su posición (`venta[1]`). Esto hace el código más legible y evita errores si cambia el orden de los datos.
* **Una lista para agrupar las ventas**, porque permite almacenar una cantidad variable de registros y recorrerlos fácilmente con un bucle `for`.

Esta estructura (lista de diccionarios) es además muy similar a la forma en que se organizan los datos en un DataFrame de Pandas, donde cada fila es un registro y cada columna tiene un nombre.

### Análisis de tendencias

Con los datos integrados se analizan dos aspectos:

1. **La evolución de las ventas a lo largo del tiempo**, agrupando la facturación por mes.
2. **La relación entre la inversión en marketing y las ventas de cada producto**, comparando el costo total de campañas con las unidades vendidas y la facturación.

````python
# Facturación total por mes
ventas_por_mes = ventas_marketing.groupby(ventas_marketing["fecha_venta"].dt.to_period("M"))["monto_venta"].sum()

ventas_por_mes
````

````python
print("Mes con mayor facturación:", ventas_por_mes.idxmax(), "-> $", round(ventas_por_mes.max(), 2))
print("Mes con menor facturación:", ventas_por_mes.idxmin(), "-> $", round(ventas_por_mes.min(), 2))
print("Facturación mensual promedio: $", round(ventas_por_mes.mean(), 2))
````

````python
# Resumen por producto: ventas e inversión en marketing
# Se usa "first" para el costo porque, después del merge, ese valor se repite en cada venta del producto
resumen_producto = ventas_marketing.groupby("producto").agg(
    unidades=("cantidad", "sum"),
    facturacion=("monto_venta", "sum"),
    costo_marketing=("costo_total", "first")
).sort_values("facturacion", ascending=False)

resumen_producto
````

````python
# Correlación entre la inversión en marketing y las ventas
correlacion = resumen_producto["costo_marketing"].corr(resumen_producto["unidades"])

print("Correlación entre costo de marketing y unidades vendidas:", round(correlacion, 2))

if abs(correlacion) < 0.3:
    print("La relación es débil: una mayor inversión en marketing no se traduce claramente en más ventas.")
elif correlacion > 0:
    print("Existe una relación positiva: los productos con más inversión en marketing tienden a vender más.")
else:
    print("Existe una relación negativa: los productos con más inversión en marketing tienden a vender menos.")
````

El resumen por producto permite comparar, para cada producto, cuánto se invirtió en marketing y cuánto se vendió. El coeficiente de correlación mide qué tan asociadas están ambas variables: valores cercanos a 1 indican que a mayor inversión hay más ventas, valores cercanos a 0 indican que no hay una relación clara, y valores cercanos a -1 indican una relación inversa.

#### Actividad 1 — Estadística descriptiva

Se calculan medidas de **tendencia central** (media, mediana y moda) y de **dispersión** (desviación estándar, varianza, rango, rango intercuartílico y coeficiente de variación). Para no repetir el mismo cálculo varias veces, se define una función que recibe una columna y devuelve todas las medidas juntas.

````python
def medidas_descriptivas(serie):
    """Devuelve las medidas de tendencia central y dispersión de una serie numérica."""
    q1 = serie.quantile(0.25)
    q3 = serie.quantile(0.75)

    # Si ningún valor se repite, la moda no tiene sentido y se deja vacía
    moda = serie.mode()
    moda = moda.iloc[0] if serie.duplicated().any() else np.nan

    return pd.Series({
        "Cantidad de datos": serie.count(),
        "Media": serie.mean(),
        "Mediana": serie.median(),
        "Moda": moda,
        "Desviación estándar": serie.std(),
        "Varianza": serie.var(),
        "Mínimo": serie.min(),
        "Máximo": serie.max(),
        "Rango": serie.max() - serie.min(),
        "Q1 (percentil 25)": q1,
        "Q3 (percentil 75)": q3,
        "Rango intercuartílico": q3 - q1,
        "Coeficiente de variación (%)": serie.std() / serie.mean() * 100,
    })
````

##### Medidas por venta individual

##### Valores atípicos

Se usa el criterio del rango intercuartílico: un valor es atípico si está más de 1,5 veces el rango intercuartílico por debajo de Q1 o por encima de Q3.

````python
def resumen_atipicos(serie):
    q1 = serie.quantile(0.25)
    q3 = serie.quantile(0.75)
    iqr = q3 - q1
    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr
    atipicos = serie[(serie < limite_inferior) | (serie > limite_superior)]
    return pd.Series({
        "Límite inferior": limite_inferior,
        "Límite superior": limite_superior,
        "Cantidad de atípicos": len(atipicos),
    })

tabla_atipicos = pd.DataFrame({
    "Precio unitario ($)": resumen_atipicos(df_ventas["precio"]),
    "Cantidad (unidades)": resumen_atipicos(df_ventas["cantidad"]),
    "Monto de la venta ($)": resumen_atipicos(df_ventas["monto_venta"]),
    "Facturación mensual ($)": resumen_atipicos(ventas_mensuales),
})

tabla_atipicos
````

##### Interpretación de la estadística descriptiva

**Por venta individual (2.998 ventas):**

* El **precio unitario** tiene una media de \$75,29 y una mediana de \$75,20, casi iguales, por lo que los precios se reparten de forma pareja entre \$26 y \$125. La moda no es representativa, porque el precio es una variable continua y casi no se repite.
* La **cantidad** por venta va de 1 a 12 unidades, con una media de 6,5 y una mediana de 7.
* El **monto de la venta** tiene una media de \$489,36 y una mediana de \$418,06. Que la media sea mayor que la mediana indica una **asimetría a la derecha**: la mayoría de las ventas son de montos bajos o medios, y unas pocas ventas grandes suben el promedio. Por eso la mediana representa mejor una venta "típica".
* El monto es la variable más dispersa: su desviación estándar es de \$334,28 y su coeficiente de variación del 68 %. El 50 % central de las ventas está entre \$220,92 (Q1) y \$709,92 (Q3).

**Ventas mensuales:** la facturación promedio por mes es de \$122.257,79, con una desviación estándar de \$10.361 y un coeficiente de variación del 8,5 %. Es decir, **las ventas mensuales son bastante estables**. El mejor mes fue mayo (\$143.727) y el peor, junio (\$108.480). La moda no aplica, porque ningún mes repite su facturación.

**Por categoría:** las tres categorías tienen una cantidad de ventas casi idéntica (alrededor de 1.000 cada una) y medianas muy parecidas (entre \$412 y \$420). Electrodomésticos tiene la media más alta (\$505,30), lo que indica que tiene algunas ventas de monto más elevado.

**Valores atípicos:** solo 9 de las 2.998 ventas (0,3 %) superan el límite superior de \$1.443,42. No son errores de carga, sino ventas grandes y poco frecuentes (cantidades altas de productos caros), por lo que se mantienen en el análisis.

#### Actividad 3 — Correlación

Se mide la relación entre variables con el **coeficiente de correlación de Pearson**, que va de -1 a 1: cerca de 1 indica que cuando una variable sube la otra también, cerca de -1 que cuando una sube la otra baja, y cerca de 0 que no hay relación lineal.

Se analiza en dos niveles: **por venta** (cada fila) y **por producto** (sumando todas sus ventas).

````python
# Correlación entre las variables de cada venta
correlacion_ventas = (df_ventas[["precio", "cantidad", "monto_venta"]].corr() + 0.0).round(2)
correlacion_ventas.index = correlacion_ventas.columns = ["Precio", "Cantidad", "Monto"]

# Correlación entre las variables de cada producto
resumen_producto["precio_promedio"] = df_ventas.groupby("producto")["precio"].mean()
correlacion_productos = resumen_producto[["precio_promedio", "unidades", "facturacion", "costo_marketing"]].corr().round(2)
correlacion_productos.index = correlacion_productos.columns = ["Precio prom.", "Unidades", "Facturación", "Marketing"]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.8))
sns.heatmap(correlacion_ventas, annot=True, cmap="RdBu", vmin=-1, vmax=1, center=0,
            square=True, cbar=False, ax=ax1, linewidths=2, linecolor="white")
ax1.set_title("Por venta")
sns.heatmap(correlacion_productos, annot=True, cmap="RdBu", vmin=-1, vmax=1, center=0,
            square=True, cbar_kws={"shrink": 0.8, "label": "Coeficiente de Pearson"},
            ax=ax2, linewidths=2, linecolor="white")
ax2.set_title("Por producto")
plt.suptitle("Matrices de correlación", fontsize=13)
plt.tight_layout()
plt.show()
````

````python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.8))

# Precio vs. cantidad en cada venta. Como la cantidad es un número entero,
# se agrega un pequeño desplazamiento aleatorio ("jitter") para que no se tapen los puntos.
r_precio_cantidad = df_ventas["precio"].corr(df_ventas["cantidad"]) + 0.0  # + 0.0 evita que se muestre "-0.00"
jitter = np.random.default_rng(0).uniform(-0.25, 0.25, len(df_ventas))
ax1.scatter(df_ventas["precio"], df_ventas["cantidad"] + jitter, s=10, alpha=0.35, color=color_principal)
pendiente, ordenada = np.polyfit(df_ventas["precio"], df_ventas["cantidad"], 1)
x = np.linspace(df_ventas["precio"].min(), df_ventas["precio"].max(), 100)
ax1.plot(x, pendiente * x + ordenada, color=color_secundario, linewidth=2)
ax1.set_title(f"Precio vs. cantidad por venta (r = {r_precio_cantidad:.2f})")
ax1.set_xlabel("Precio unitario ($)")
ax1.set_ylabel("Cantidad vendida")

# Inversión en marketing vs. unidades vendidas por producto
r_marketing = resumen_producto["costo_marketing"].corr(resumen_producto["unidades"])
categoria_producto = df_ventas.groupby("producto")["categoria"].first()
ax2.scatter(resumen_producto["costo_marketing"], resumen_producto["unidades"], s=60,
            color=categoria_producto.reindex(resumen_producto.index).map(colores_categoria),
            edgecolor="white", linewidth=1.5)
pendiente, ordenada = np.polyfit(resumen_producto["costo_marketing"], resumen_producto["unidades"], 1)
x = np.linspace(resumen_producto["costo_marketing"].min(), resumen_producto["costo_marketing"].max(), 100)
ax2.plot(x, pendiente * x + ordenada, color=color_secundario, linewidth=2)
ax2.set_title(f"Inversión en marketing vs. unidades por producto (r = {r_marketing:.2f})")
ax2.set_xlabel("Costo total de campañas")
ax2.set_ylabel("Unidades vendidas")
from matplotlib.patches import Patch
ax2.legend(handles=[Patch(color=c, label=n) for n, c in colores_categoria.items()], frameon=False)

plt.tight_layout()
plt.show()

print("Correlación precio - cantidad (por venta):", f"{r_precio_cantidad:.3f}")
print("Correlación precio promedio - unidades (por producto):", correlacion_productos.loc["Precio prom.", "Unidades"])
print("Correlación marketing - unidades (por producto):", round(r_marketing, 2))
````

##### Interpretación de las correlaciones

* **Precio y cantidad (r = 0,00):** no hay ninguna relación entre el precio de un producto y la cantidad que se compra en cada venta. Los clientes compran cantidades similares sin importar si el producto es barato o caro. El resultado es el mismo dentro de cada categoría.
* **Monto con cantidad (r = 0,77) y con precio (r = 0,56):** el monto de una venta depende más de **cuántas unidades** se compran que del precio unitario. Para aumentar la facturación, sería más efectivo impulsar la compra de más unidades (promociones por cantidad, combos) que vender productos más caros.

Por ejemplo, Alfombra y Cortinas están entre los productos con más inversión y venden menos que el promedio.
* **Unidades y facturación por producto (r = 0,99):** casi perfecta, como es esperable, porque todos los productos tienen precios promedio similares (entre \$71 y \$81).

Hay que recordar que **correlación no implica causalidad**. Además, la inversión en marketing es muy baja en relación con las ventas (entre \$11,85 y \$18,16 por producto en todo el año), lo que limita el efecto que podría tener.

#### Actividad 3 — Visualización interactiva con Plotly

Se construye un **dashboard interactivo** que combina cuatro visualizaciones. Con el menú desplegable de arriba a la izquierda se puede filtrar por categoría, y todos los gráficos y los indicadores del título se actualizan juntos. Además, cada gráfico permite pasar el mouse para ver los valores exactos, hacer zoom y ocultar series desde la leyenda.

````python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

filtros = ["Todas"] + orden_categorias
color_filtro = {"Todas": color_principal, **colores_categoria}

fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=("Facturación mensual", "Top 10 productos por facturación",
                    "Facturación por día de la semana", "Precio vs. cantidad por venta"),
    vertical_spacing=0.14, horizontal_spacing=0.1,
)

titulos = {}
for filtro in filtros:
    datos = df_ventas if filtro == "Todas" else df_ventas[df_ventas["categoria"] == filtro]
    color = color_filtro[filtro]
    visible = filtro == "Todas"

    # 1. Facturación mensual (línea)
    mensual = datos.groupby(datos["fecha_venta"].dt.month)["monto_venta"].sum()
    fig.add_trace(go.Scatter(
        x=[nombres_meses[m - 1] for m in mensual.index], y=mensual.values, mode="lines+markers",
        line=dict(color=color, width=2), marker=dict(size=8), name=filtro, visible=visible,
        hovertemplate="%{x}: $%{y:,.0f}<extra></extra>"), row=1, col=1)

    # 2. Top 10 productos (barras horizontales)
    top = datos.groupby("producto")["monto_venta"].sum().nlargest(10).sort_values()
    fig.add_trace(go.Bar(
        x=top.values, y=top.index, orientation="h", marker_color=color, name=filtro, visible=visible,
        hovertemplate="%{y}: $%{x:,.0f}<extra></extra>"), row=1, col=2)

    # 3. Facturación por día de la semana (barras)
    dia = datos.groupby(datos["fecha_venta"].dt.dayofweek)["monto_venta"].sum()
    fig.add_trace(go.Bar(
        x=[dias_semana[d] for d in dia.index], y=dia.values, marker_color=color, name=filtro, visible=visible,
        hovertemplate="%{x}: $%{y:,.0f}<extra></extra>"), row=2, col=1)

    # 4. Precio vs. cantidad (dispersión)
    fig.add_trace(go.Scatter(
        x=datos["precio"], y=datos["cantidad"], mode="markers", name=filtro, visible=visible,
        marker=dict(color=color, size=6, opacity=0.35),
        customdata=datos[["producto", "monto_venta"]],
        hovertemplate="%{customdata[0]}<br>Precio: $%{x:.2f}<br>Cantidad: %{y}<br>Monto: $%{customdata[1]:,.2f}<extra></extra>"),
        row=2, col=2)

    # Indicadores que se muestran en el título según el filtro
    titulos[filtro] = (f"<b>Dashboard de ventas 2024 — {filtro}</b><br>"
                       f"<span style='font-size:13px'>Facturación: ${datos['monto_venta'].sum():,.0f}  ·  "
                       f"Ventas: {len(datos):,}  ·  Unidades: {datos['cantidad'].sum():,}  ·  "
                       f"Ticket promedio: ${datos['monto_venta'].mean():,.2f}</span>")

# Menú desplegable: cada opción muestra solo las 4 series de su categoría
trazas_por_filtro = 4
botones = []
for i, filtro in enumerate(filtros):
    visibles = [False] * (len(filtros) * trazas_por_filtro)
    visibles[i * trazas_por_filtro:(i + 1) * trazas_por_filtro] = [True] * trazas_por_filtro
    botones.append(dict(label=filtro, method="update",
                        args=[{"visible": visibles}, {"title.text": titulos[filtro]}]))

fig.update_layout(
    title=dict(text=titulos["Todas"], x=0.5),
    updatemenus=[dict(buttons=botones, direction="down", x=0, xanchor="left", y=1.17, yanchor="top",
                      showactive=True, bgcolor="white")],
    annotations=list(fig.layout.annotations) + [dict(text="Categoría:", x=0, xref="paper", y=1.21,
                                                     yref="paper", showarrow=False, xanchor="left")],
    height=820, showlegend=False, template="plotly_white",
    margin=dict(t=150, l=60, r=30, b=50),
)
fig.update_yaxes(tickprefix="$", tickformat=",.0f", row=1, col=1)
fig.update_xaxes(tickprefix="$", tickformat=",.0f", row=1, col=2)
fig.update_yaxes(tickprefix="$", tickformat=",.0f", row=2, col=1)
fig.update_xaxes(title_text="Precio unitario ($)", row=2, col=2)
fig.update_yaxes(title_text="Cantidad", row=2, col=2)

fig.show()
````

Como complemento, se agrega un gráfico de la facturación diaria con **selector de período**: los botones permiten ver el último mes, los últimos 3 o 6 meses o todo el año, y la barra inferior permite elegir cualquier rango de fechas arrastrando.

````python
diaria_categoria = df_ventas.groupby(["fecha_venta", "categoria"])["monto_venta"].sum().unstack().fillna(0)

fig2 = go.Figure()
for categoria in orden_categorias:
    fig2.add_trace(go.Scatter(
        x=diaria_categoria.index, y=diaria_categoria[categoria].rolling(7).mean(),
        name=categoria, mode="lines", line=dict(color=colores_categoria[categoria], width=2),
        hovertemplate="%{x|%d/%m/%Y}: $%{y:,.0f}<extra>" + categoria + "</extra>"))

fig2.update_layout(
    title="Facturación diaria por categoría (media móvil de 7 días)",
    template="plotly_white", height=480, hovermode="x unified",
    yaxis=dict(tickprefix="$", tickformat=",.0f"),
    xaxis=dict(
        rangeselector=dict(buttons=[
            dict(count=1, label="1 mes", step="month", stepmode="backward"),
            dict(count=3, label="3 meses", step="month", stepmode="backward"),
            dict(count=6, label="6 meses", step="month", stepmode="backward"),
            dict(step="all", label="Todo el año"),
        ]),
        rangeslider=dict(visible=True),
        type="date",
    ),
    legend=dict(orientation="h", y=1.12, x=1, xanchor="right"),
)
fig2.show()
````

##### Cómo usar el dashboard

* **Filtro por categoría:** el menú desplegable permite elegir "Todas" o una categoría. Al cambiarlo, se actualizan los cuatro gráficos y los indicadores del título (facturación, ventas, unidades y ticket promedio).
* **Hover:** al pasar el mouse sobre cualquier punto o barra se ve el valor exacto. En el gráfico de dispersión se muestra además el producto y el monto de esa venta.
* **Zoom:** se puede arrastrar sobre cualquier gráfico para acercarse, y hacer doble clic para volver a la vista completa.
* **Selector de período:** en el gráfico de facturación diaria, los botones permiten ver el último mes, los últimos 3 o 6 meses o el año completo, y la barra inferior permite elegir cualquier rango de fechas.

A diferencia de los gráficos de Matplotlib y Seaborn, que son imágenes fijas, el dashboard permite que cada persona explore los datos según la pregunta que tenga, sin necesidad de escribir código.
